# Loading microsft - Phi3 (3.8B)

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

model = AutoModelForCausalLM.from_pretrained(
    "microsoft/phi-3-mini-4k-instruct",
    device_map="cuda",
    torch_dtype="auto",
    trust_remote_code=True,
)

tokenizer = AutoTokenizer.from_pretrained("microsoft/phi-3-mini-4k-instruct")

#Create a pipeline
pipe =pipeline("text-generation",
               model=model,
               tokenizer=tokenizer,
               return_full_text=False,
               max_new_tokens=500,
               do_sample=False,
               )
```

```python
#Prompt
messages =[
    {"role":"user","content":"Create a funny joke about spiderman."}
]
output =pipe(messages)
print(output[0]["generated_text"])
output =pipe(messages,do_sample=True,temperature=1)
print(output[0]["generated_text"])
```

```python
#Apply Prompt template
prompt =pipe.tokenizer.apply_chat_template(messages, tokenize=False)
print(prompt)
```

# In Context_learning has 3 types 0-shot,1-shot,few-shot prompting techniques

```python
# Using one-shot prompt

one_shot_prompt =[
    {
    "role":"user",
    "content":"A 'Piramaru' is a Persian Musical Instrument. An example of sentence using the word Piramaru is"
    },
    {
        "role":"assistant",
        "content":"I have a Piramaru that my Persian Friend gave me as a gift. I love to play it all the time"
    },
    {"role":"user",
    "content": "'laagadam' is an act of snatching away quickly. An example of sentence using the word laagadam is"}
]
print(tokenizer.apply_chat_template(one_shot_prompt, tokenize=False))
```

```python
outputs =pipe(one_shot_prompt)
print(outputs[0]["generated_text"])
output =pipe(one_shot_prompt,do_sample=True,temperature=1)
print(output[0]["generated_text"])
```
